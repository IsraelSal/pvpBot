<h3>Search a pdf files appartamenti per le prossime aste

```python
def page_load_id(url_data):
    browser.get(url_data)
    div_link = browser.find_elements(By.TAG_NAME, "a")
    links = [d.get_attribute("href") for d in div_link]
    valid_links = [d for d in links if d != None]
    link_asta = [s for s in valid_links if s.find(".pdf")!=-1]
    return link_asta
```
