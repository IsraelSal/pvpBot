download i link validi

```python
import numpy as np

div_link = browser.find_elements(By.TAG_NAME, "a")
links = [d.get_attribute("href") for d in div_link]
valid_links = [d for d in links if d != None]
link_asta = [s for s in valid_links if s.find("contentId")!=-1]
unique_ = np.unique(np.array(link_asta))
unique_.shape
```

i 7 appartamenti per le prossime aste

```python
from time import sleep
last7 = unique_[1:7]
action = ActionChains(browser)

for url_i in last7:
    str_compose = f'''
    window.open("{url_i}","_blank");
    '''
    browser.execute_script(str_compose)
    sleep(2)
```
