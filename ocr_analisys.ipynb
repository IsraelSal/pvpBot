<h3>Load the tesseract engine

```python
pytesseract.pytesseract.tesseract_cmd = "C:/Program Files/Tesseract-OCR/tesseract.exe"
```

Make a screenshot and Split the pdf by page 

```python
pdfs = r"Test.pdf"
pages = convert_from_path(pdfs, 350, poppler_path="C:/Users/ADMIN/poppler-0.68.0/bin")

i = 1
for page in pages:
    image_name = "Page_" + str(i) + ".jpg"  
    page.save(image_name, "JPEG")
    i = i+1   
```

<h2>Convert image in text 

```python

# Adding custom options
custom_config = r'--oem 3 --psm 6'
perizia = ""
for i in range(1,16):

    img = cv2.imread(f'Page_{i}.jpg')

    image_ = pytesseract.image_to_string(img, config=custom_config)
    perizia = perizia + "\n" + image_ 
```

<h3> Save like text file

```python
with open("data/perizia.txt", 'w') as d:
    d.write(perizia)
```

## Now we try to read the vectorial PDF and extract the text

```python
pdffileobj=open('Test.pdf','rb')
pdfreader=PyPDF2.PdfFileReader(pdffileobj)
x=pdfreader.numPages
x
```

```python
testo_comp = ""
for i in range(0, x):
    pageobj=pdfreader.getPage(i)
    text=pageobj.extractText()
    testo_comp += text
```

```python
to_save = open("data/perizia_vett.txt","w")
to_save.writelines(testo_comp)
```
